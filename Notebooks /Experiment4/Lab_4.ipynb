# Experiment 4: Term Frequency and Named Entity Recognition

### 4.1 Toolkit-based TF & NER

In this section, we leverage a pre-trained NLP toolkit, spaCy, to efficiently process text. We will:

1.  **Load** an English language model.
2.  **Read** our input text from `4.1_4.2_input.txt`.
3.  **Process** the text to extract tokens and entities.
4.  **Calculate Term Frequencies** for all words, filtering out punctuation and whitespace.
5.  **Identify** and display the top 10 most frequent terms.
6.  **Save** the complete term frequency distribution to a CSV file.
7.  **Perform Named Entity Recognition (NER)** to find and categorize entities like persons, organizations, and locations, displaying the top 10 found.

```python
import sys

# Clear a partially imported pandas package left by a previous failed run
for module_name in list(sys.modules):
    if module_name == 'pandas' or module_name.startswith('pandas.'):
        del sys.modules[module_name]

import spacy
import pandas as pd
from collections import Counter
from pathlib import Path

print('Libraries imported successfully.')
```

#### Load spaCy Model and Input Data

First, we need to load a spaCy English language model. If it's not already downloaded, the code will attempt to download it. Then, we'll read the text from our specified input file, `4.1_4.2_input.txt`.

```python
# Load spaCy English model
try:
    nlp = spacy.load('en_core_web_sm')
    print('spaCy model \'en_core_web_sm\' loaded.')
except OSError:
    print('spaCy model \'en_core_web_sm\' not found. Attempting download...')
    import spacy.cli
    spacy.cli.download('en_core_web_sm')
    nlp = spacy.load('en_core_web_sm')
    print('spaCy model \'en_core_web_sm\' downloaded and loaded.')

# Read input data from the notebook folder or the workspace root
input_path = Path('4.1_4.2_input.txt')
if not input_path.exists():
    # Fallback to a common directory name if the file isn't in the root
    input_path = Path('Experiment4') / '4.1_4.2_input.txt'

if not input_path.exists():
    print(f"Error: Input file '{input_path}' not found. Please ensure '4.1_4.2_input.txt' is in the current directory or 'Experiment4' subfolder.")
    text = ""
else:
    text = input_path.read_text(encoding='utf-8')
    print(f'Text loaded from: {input_path}')

# Process with spaCy
doc = nlp(text)
print('Text processed by spaCy.')
```

#### Calculate Term Frequencies

We'll extract individual words (tokens) from the processed document, convert them to lowercase, and filter out punctuation and whitespace. Then, we use Python's `collections.Counter` to count the occurrences of each word, providing us with their frequencies. We'll display the top 10 most frequent terms and save the full list to a CSV file.

```python
# Extract Term Frequency (Filtering out punctuation and whitespace)
words = [token.text.lower() for token in doc if not token.is_punct and not token.is_space and token.text.strip() != '']
word_freq = Counter(words)

# Identify top 10 most frequent terms
top_10_words = word_freq.most_common(10)
print('Top 10 most frequent terms (Toolkit):')
for term, freq in top_10_words:
    print(f'{term}: {freq}')

# Save to CSV
df_tf_toolkit = pd.DataFrame(word_freq.items(), columns=['Term', 'Frequency'])
df_tf_toolkit.to_csv('4.1_TF_Toolkit.csv', index=False)
print('\nTerm frequencies saved to 4.1_TF_Toolkit.csv')
```

#### Named Entity Recognition (NER)

Finally, we'll use spaCy's built-in NER capabilities to identify and classify named entities within the text. Entities can include names of people, organizations, locations, dates, and more. We'll display the first 10 entities found.

```python
# Named Entity Recognition (NER)
print('\nNamed Entities found (Top 10):')
entities = [(ent.text, ent.label_) for ent in doc.ents]
df_ner = pd.DataFrame(entities, columns=['Entity', 'Label'])
display(df_ner.head(10))
```

### 4.2 Native Term-Frequency Analysis (Without Toolkit)

Now, let's perform the same term frequency analysis using only basic Python constructs, without relying on specialized NLP libraries like spaCy. This 'from-scratch' approach helps us understand the fundamental steps involved in text processing.

We will:

1.  **Read** the same input text (`4.1_4.2_input.txt`).
2.  **Preprocess** the text by converting it to lowercase and removing punctuation.
3.  **Tokenize** the text by splitting it into words.
4.  **Calculate Term Frequencies** manually using a Python dictionary.
5.  **Identify** and display the top 10 most frequent terms.
6.  **Save** the complete term frequency distribution to a CSV file.

```python
import string
import csv
from pathlib import Path

print('Libraries imported for Native TF analysis.')
```

#### Load Data and Preprocess

We start by loading the input file again. Then, we'll implement a basic preprocessing pipeline: converting all text to lowercase and removing all punctuation. This ensures that words like 'Word' and 'word.' are treated as the same token ('word').

```python
# Read input data from the notebook folder or the workspace root
input_path = Path('4.1_4.2_input.txt')
if not input_path.exists():
    input_path = Path('Experiment4') / '4.1_4.2_input.txt'

if not input_path.exists():
    print(f"Error: Input file '{input_path}' not found. Please ensure '4.1_4.2_input.txt' is in the current directory or 'Experiment4' subfolder.")
    text = ""
else:
    text = input_path.read_text(encoding='utf-8')
    print(f'Text loaded from: {input_path}')

# Preprocessing: lowercase and remove punctuation
text_lower = text.lower()
translator = str.maketrans('', '', string.punctuation)
clean_text = text_lower.translate(translator)
print('Text preprocessed (lowercased and punctuation removed).')

# Tokenize by splitting on whitespace
tokens = clean_text.split()
print(f'Text tokenized into {len(tokens)} words.')
```

#### Calculate Native Term Frequencies

Here, we manually count word occurrences using a dictionary. Each word becomes a key, and its count is the value. After calculating all frequencies, we sort them to find the top 10 most frequent terms and save the entire distribution to a CSV.

```python
# Calculate Term Frequency natively
native_word_freq = {}
for token in tokens:
    # Only count non-empty strings (after splitting and removing punctuation, some might be empty)
    if token:
        native_word_freq[token] = native_word_freq.get(token, 0) + 1

# Identify top 10 most frequent terms
sorted_freq = sorted(native_word_freq.items(), key=lambda x: x[1], reverse=True)
top_10_native = sorted_freq[:10]

print('Top 10 most frequent terms (Native):')
for term, freq in top_10_native:
    print(f'{term}: {freq}')

# Save to CSV natively
with open('4.2_TF_Native.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Term', 'Frequency'])
    writer.writerows(native_word_freq.items())

print('\nTerm frequencies saved to 4.2_TF_Native.csv')
```

### 4.3 Native TF-IDF Computation

This section delves into a more advanced text analysis technique: Term Frequency-Inverse Document Frequency (TF-IDF). TF-IDF is a numerical statistic that reflects how important a word is to a document in a collection or corpus. It is often used as a weighting factor in information retrieval and text mining.

We will implement TF-IDF from scratch using only Python's standard library. The process involves:

1.  **Preprocessing and Tokenizing** a small set of sample documents.
2.  **Calculating Term Frequency (TF)** for each term within each document.
3.  **Calculating Document Frequency (DF)** for each term across the entire set of documents.
4.  **Deriving Inverse Document Frequency (IDF)** using a logarithmic formula.
5.  **Computing the final TF-IDF scores** for every term in every document.
6.  **Displaying** the results, highlighting the top 10 terms with the highest TF-IDF scores for each document.

```python
import math
import string

print('Libraries imported for Native TF-IDF computation.')
```

#### Define Documents and Preprocessing Function

We start by defining a small corpus of sample documents. We also create a simple `preprocess` function that converts text to lowercase, removes punctuation, and splits it into words. This ensures consistent tokenization across all documents.

```python
# Given Documents
docs = [
    'Natural language processing is a field of artificial intelligence.',
    'Natural language processing helps computers understand human language.',
    'Machine learning is an important part of artificial intelligence.'
]

def preprocess(doc):
    doc = doc.lower()
    doc = doc.translate(str.maketrans('', '', string.punctuation))
    return doc.split()

# 1. Preprocess and Tokenize all documents
tokenized_docs = [preprocess(doc) for doc in docs]
print(f'Processed {len(tokenized_docs)} documents.')
```

#### Calculate Term Frequency (TF)

Term Frequency (TF) measures how frequently a term appears in a document. Since documents of different lengths might have different word counts, we normalize TF by dividing the number of times a term appears in a document by the total number of terms in that document. This gives us the relative frequency.

```python
# 2. Calculate TF for each term in each document
tf_docs = []
for tokens in tokenized_docs:
    tf = {}
    total_words = len(tokens)
    if total_words == 0: # Handle empty documents to avoid division by zero
        tf_docs.append(tf)
        continue

    for word in tokens:
        tf[word] = tf.get(word, 0) + 1
    for word in tf:
        tf[word] = tf[word] / total_words  # Normalized Term Frequency
    tf_docs.append(tf)
print('Term Frequencies (TF) calculated for all documents.')
```

#### Calculate Document Frequency (DF) and Inverse Document Frequency (IDF)

**Document Frequency (DF)** counts the number of documents in the corpus that contain a given term.

**Inverse Document Frequency (IDF)** measures how important a term is. While TF increases with the number of times a word appears in a document, IDF is high for rare words and low for common words (like 'the', 'is', 'a') that appear in many documents. The formula used is `log(Total Documents / (1 + Document Frequency))`. We add 1 to the DF in the denominator to avoid division by zero for terms not present in any document (though in this specific implementation, terms not present won't have an entry in `df`).

```python
# 3. Calculate DF for every term across the corpus
df = {}
for tokens in tokenized_docs:
    unique_terms = set(tokens) # Get unique terms for each document
    for term in unique_terms:
        df[term] = df.get(term, 0) + 1
print('Document Frequencies (DF) calculated.')

# 4. Calculate IDF using logarithmic formula
total_docs = len(docs)
idf = {}
for term, freq in df.items():
    # Add 1 to denominator to avoid division by zero for unseen terms (though all terms here are seen)
    idf[term] = math.log(total_docs / freq)
print('Inverse Document Frequencies (IDF) calculated.')
```

#### Compute TF-IDF Scores and Display Results

Finally, the TF-IDF score for a term in a document is simply the product of its TF and IDF values. A high TF-IDF score suggests that the term is important in that specific document and relatively rare across the entire corpus. We will calculate these scores and then display them for each document, along with the top 10 terms that have the highest TF-IDF scores.

```python
# 5. Calculate TF-IDF for every term in every document
tfidf_docs = []
for tf_dict in tf_docs:
    tfidf = {}
    for term, tf_value in tf_dict.items():
        tfidf[term] = tf_value * idf[term]
    tfidf_docs.append(tfidf)
print('TF-IDF scores calculated for all documents.')

# 6. Display results
for i, doc_tfidf in enumerate(tfidf_docs):
    print(f'\n--- Document {i+1} ---')
    print(f'{"Term":<15} | {"TF":<8} | {"DF":<2} | {"IDF":<8} | {"TF-IDF"}')
    print('-' * 55)

    # Display each term's individual TF, DF, IDF, and TF-IDF values for clarity
    # Only display once per term, even if it appears multiple times in the document
    displayed = set()
    for term in tokenized_docs[i]:
        if term not in displayed:
            t_tf = tf_docs[i].get(term, 0.0) # Use .get to handle potential missing keys gracefully
            t_df = df.get(term, 0)
            t_idf = idf.get(term, 0.0)
            t_tfidf = doc_tfidf.get(term, 0.0)
            print(f'{term:<15} | {t_tf:<8.4f} | {t_df:<2} | {t_idf:<8.4f} | {t_tfidf:.4f}')
            displayed.add(term)

    # Identify the top 10 terms with the highest TF-IDF score for this document
    top_10 = sorted(doc_tfidf.items(), key=lambda x: x[1], reverse=True)[:10]
    print('\nTop Terms (Highest TF-IDF):')
    for term, score in top_10:
        print(f'{term}: {score:.4f}')
```
