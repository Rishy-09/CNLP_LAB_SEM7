# Experiment 3: Text Normalization and Information Extraction

This experiment focuses on fundamental Natural Language Processing (NLP) techniques: text normalization (stemming and lemmatization) and information extraction using regular expressions. These techniques are crucial for preparing text data for further analysis and for extracting meaningful insights from unstructured text.

## Experiment 3.1: Stemming using Porter Stemmer

Stemming is a technique to reduce inflected (or sometimes derived) words to their word stem, base, or root form—generally a written word form. The stem is not necessarily a morphological root of the word; it is usually an affix, a common form to which all words are reduced. Here, we'll use the Porter Stemmer, a popular algorithm for English.

```python
# Install NLTK if you haven't already
# !pip install nltk

import nltk
from nltk.stem import PorterStemmer

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('stemmers/porter')
except LookupError:
    nltk.download('porter_stemmer')

stemmer = PorterStemmer()
words = ['playing', 'played', 'plays', 'studies', 'studying', 'connected', 'connection', 'computers']

print(f"{'Original Word':<15} -> {'Stem'}")
print("-" * 30)
for word in words:
    print(f"{word:<15} -> {stemmer.stem(word)}")
```

As you can observe, the Porter Stemmer effectively reduces various forms of words like 'playing', 'played', 'plays' to 'play', and 'connected', 'connection' to 'connect'. However, it can sometimes produce stems that are not actual English words, such as 'studi' from 'studies' and 'studying', or 'comput' from 'computers'. This is characteristic of rule-based stemming algorithms.

## Experiment 3.2: Lemmatization using WordNet Lemmatizer

Lemmatization, unlike stemming, aims to reduce words to their base or root form, known as a lemma, which is a dictionary word. It considers the word's morphological analysis and part-of-speech to return a valid word. We'll use the WordNet Lemmatizer, which requires Part-of-Speech (POS) tags for better accuracy.

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')
try:
    nltk.data.find('corpora/omw-1.4')
except LookupError:
    nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
words = ['cats', 'dogs', 'running', 'runs', 'ran', 'studies', 'studying', 'better', 'children', 'mice', 'went', 'ate', 'leaves', 'caring']

# Manually assign POS tags for demonstration. In a real scenario, these would be predicted.
pos_tags = {
    'cats': wordnet.NOUN, 'dogs': wordnet.NOUN, 'running': wordnet.VERB,
    'runs': wordnet.VERB, 'ran': wordnet.VERB, 'studies': wordnet.VERB,
    'studying': wordnet.VERB, 'better': wordnet.ADJ, 'children': wordnet.NOUN,
    'mice': wordnet.NOUN, 'went': wordnet.VERB, 'ate': wordnet.VERB,
    'leaves': wordnet.NOUN, 'caring': wordnet.VERB
}

print(f"{'Original Word':<15} -> {'Lemma'}")
print("-" * 30)
for word in words:
    # Using the appropriate POS tag for accurate lemmatization
    lemma = lemmatizer.lemmatize(word, pos=pos_tags[word])
    print(f"{word:<15} -> {lemma}")
```

Lemmatization provides more accurate base forms compared to stemming because it takes into account the word's meaning and part of speech. For instance, 'studies' and 'studying' are both reduced to the verb 'study', 'better' (adjective) becomes 'good', and irregular forms like 'went' become 'go'. This makes lemmatization more suitable for applications where semantic accuracy is crucial.

## Experiment 3.3: Information Extraction using Regular Expressions (Regex)

Regular Expressions (Regex) are powerful tools for pattern matching and information extraction from text. They allow us to define specific patterns to search for and extract desired data points, such as email addresses, URLs, mobile numbers, hashtags, and mentions. This experiment demonstrates how to use `re` module in Python to extract these entities from a given text.

```python
import re

text = """Welcome to the Natural Language Processing workshop! For more information, visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your experience using #NLP, #Python, and #MachineLearning. You can also visit https://github.com/NLPWorkshop for the latest updates."""

# Regex patterns for different entities
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
url_pattern = r'https?://[^\s]+|\bwww\.[^\s]+\b'
mobile_pattern = r'(?:\+91[-\s]?)?[6-9]\d{9}' # Matches +91-XXXXXXXXXX, +91 XXXXXXXXXX, or XXXXXXXXXX
hashtag_pattern = r'#[A-Za-z0-9_]+'
mention_pattern = r'(?<!\S)@[A-Za-z0-9_]+' # Ensures @ is not preceded by a non-whitespace character

print("Email addresses :", re.findall(email_pattern, text))
# Remove trailing punctuation from URLs if any
print("URLs            :", [u.rstrip('.,') for u in re.findall(url_pattern, text)])
print("Mobile numbers  :", re.findall(mobile_pattern, text))
print("Hashtags        :", re.findall(hashtag_pattern, text))
print("Mentions        :", re.findall(mention_pattern, text))
```

This section demonstrates the precision of regular expressions in identifying and extracting specific patterns from text. Each regex pattern is designed to capture a particular type of entity (email, URL, mobile number, hashtag, mention). The output clearly shows the successful extraction of all target information, highlighting the utility of regex in tasks like data cleaning, entity recognition, and information retrieval.

## Summary of Learnings

- **Stemming vs. Lemmatization**: Stemming is a heuristic process that chops off word endings, often resulting in non-dictionary words (e.g., 'studi'). Lemmatization, on the other hand, is a more sophisticated process that uses a vocabulary and morphological analysis to return the base or dictionary form of a word (e.g., 'go' from 'went', 'good' from 'better'). Lemmatization requires Part-of-Speech context for higher accuracy.

- **Regular Expressions**: Regex provides a powerful and flexible way to extract structured information from unstructured text based on defined patterns. It is highly effective for tasks like identifying email addresses, URLs, phone numbers, and social media tags.
