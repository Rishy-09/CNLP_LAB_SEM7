## Experiment 2.1: Basic Text Preprocessing
Write a Python program to perform basic text preprocessing on a given data. The program should:
1. Convert text to lowercase. (Also display how many uppercase characters were converted)
2. Remove punctuation. (Also display how many different types of punctuations were removed)
3. Remove numbers. (Also display how many numbers were removed)
4. Remove extra whitespace. (Also display how much extra whitespace was present)
5. Display the cleaned text.


### Experiment 2.1 Implementation

```python
# Step 1: Read the content from the file
import os

file_name = '2.1_text_data.txt'

with open(file_name, 'r') as f:
    original_text = f.read()

print("Original Text:")
print(original_text)
print("\n" + "="*50 + "\n")
```

```python
# Step 2: Convert text to lowercase and count uppercase characters

uppercase_count = sum(1 for char in original_text if char.isupper())
text_lower = original_text.lower()

print(f"Number of uppercase characters converted to lowercase: {uppercase_count}")
print("Text after converting to lowercase:\n")

print(text_lower)
print("\n" + "="*50 + "\n")
```

```python
# Step 3: Remove punctuation and count different types of punctuations
import string

punctuation_chars = set()
for char in text_lower:
  # 'string.punctuation' is a string that contains -> !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
    if char in string.punctuation:
        punctuation_chars.add(char)

punctuation_count = len(punctuation_chars)

text_no_punctuation = ''.join([char for char in text_lower if char not in string.punctuation])

print(f"Number of different types of punctuations removed: {punctuation_count}")
print(f"Punctuation characters removed: {' '.join(sorted(list(punctuation_chars))) if punctuation_chars else 'None'}")
print("Text after removing punctuation:")
print(text_no_punctuation)
print("\n" + "="*50 + "\n")
```

```python
# Step 4: Remove numbers and count numbers

number_count = sum(1 for char in text_no_punctuation if char.isdigit())
text_no_numbers = ''.join([char for char in text_no_punctuation if not char.isdigit()])

print(f"Number of numeric characters removed: {number_count}")
print("Text after removing numbers:")
print(text_no_numbers)
print("\n" + "="*50 + "\n")
```

```python
# Step 5: Remove extra whitespace and count extra whitespace
import re

# Count original whitespace for comparison
original_whitespace_count = sum(1 for char in text_no_numbers if char.isspace())

# Replace multiple spaces with a single space and strip leading/trailing whitespace
text_no_extra_whitespace = re.sub(r'\s+', ' ', text_no_numbers).strip()
# strip(): remove spaces at the beginning or end of the text..
# Count new whitespace
cleaned_whitespace_count = sum(1 for char in text_no_extra_whitespace if char.isspace())

extra_whitespace_removed = original_whitespace_count - cleaned_whitespace_count

print(f"Extra whitespace characters removed: {extra_whitespace_removed}")
print("Text after removing extra whitespace:")
print(text_no_extra_whitespace)
print("\n" + "="*50 + "\n")
```

```python
# Step 6: Display the cleaned text

cleaned_text = text_no_extra_whitespace

print("Final Cleaned Text:")
print(cleaned_text)
print("\n" + "="*50 + "\n")
```

## Stemming and Lemmatization

When working with natural language, understanding the intended meaning behind words is crucial. To achieve this, we often reduce words to their root or base form, a process known as **canonicalization**.

Two primary techniques for canonicalization are:

### 2.1 Stemming
*   **Goal**: Generates the base word by removing affixes (prefixes/suffixes).
*   **Method**: Uses pre-defined rules to drop affixes.
*   **Outcome**: Might not always result in semantically meaningful base words.
*   **Performance**: Faster and computationally less expensive than lemmatization.

### 2.2 Lemmatization
*   **Goal**: Groups together inflected forms of a word to reach its base form (lemma).
*   **Method**: Uses vocabulary and morphological analysis.
*   **Outcome**: Always results in a semantically meaningful base form (lemma).
*   **Performance**: Slower and computationally more expensive than stemming.

### Experiment 2.2 Implementation

```python
# Step 1: Read the content from the file for Experiment 2.2
import os

file_name_2_2 = '2.2_tokenization_data.txt'

with open(file_name_2_2, 'r') as f:
    tokenization_text = f.read()

print("Text for Tokenization:")
print(tokenization_text)
print("\n" + "="*50 + "\n")
```

#### 2.2.1 Word and Sentence Tokenization using NLTK

```python
import nltk

# Download punkt
try:
    # tokenizer data package..
    nltk.data.find('tokenizers/punkt')
except LookupError:
    # punkt: pre-trained ml model
    nltk.download('punkt')

# Download punkt_tab required by newer NLTK versions
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

print("NLTK setup complete.")
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

# Word Tokenization
nltk_words = word_tokenize(tokenization_text)
print("NLTK Word Tokens:")
print(nltk_words)

# Sentence Tokenization
nltk_sentences = sent_tokenize(tokenization_text)
print("\nNLTK Sentence Tokens:")
print(nltk_sentences)
print("\n" + "="*50 + "\n")
```

#### 2.2.2 Word and Sentence Tokenization using spaCy

```python
# Install spaCy if not already installed
try:
    import spacy
except ImportError:
    !pip install spacy
    import spacy

# Download a spaCy language model (e.g., 'en_core_web_sm')
try:
    spacy.load('en_core_web_sm')
except OSError:
    !python -m spacy download en_core_web_sm

print("spaCy setup complete.")
```

```python
import spacy

# Load the English model
nlp = spacy.load('en_core_web_sm')

# Process the text with spaCy
doc = nlp(tokenization_text)

# Word Tokenization (spaCy considers tokens as individual words/punctuation)
spacy_words = [token.text for token in doc]
print("spaCy Word Tokens:")
print(spacy_words)

# Sentence Tokenization
spacy_sentences = [sent.text for sent in doc.sents]
print("\nspaCy Sentence Tokens:")
print(spacy_sentences)
print("\n" + "="*50 + "\n")
```

#### 2.2.3 Word and Sentence Tokenization without NLTK/spaCy

```python
import re

# --- Word Tokenization (basic approach) ---
# Convert to lowercase, then split by spaces and remove non-alphanumeric characters
# This is a very basic approach and might not handle all cases perfectly.
cleaned_text_for_word_token = re.sub(r'[^\w\s]', '', tokenization_text.lower())
custom_words = cleaned_text_for_word_token.split()
print("Custom Word Tokens:")
print(custom_words)

# --- Sentence Tokenization (basic approach) ---
# Split by common sentence-ending punctuation marks followed by a space
custom_sentences = re.split(r'(?<=[.!?])\s+', tokenization_text.strip())
# `?<=` is a **positive lookbehind** in regular expressions, which checks if a
#  pattern exists directly before the current position without including that pattern in the match.
print("\nCustom Sentence Tokens:")
print(custom_sentences)
print("\n" + "="*50 + "\n")
```
