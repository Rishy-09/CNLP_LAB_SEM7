# Data Analysis on Employee Dataset

This notebook will perform various data analysis experiments on the `employee_information_100.csv` dataset.

## Experiment Objectives:

1.  **Experiment 1.1:** Read the employee dataset and display the average salary of each department using a bar chart.
2.  **Experiment 1.2:** Display the number of employees working in each department.
3.  **Experiment 1.3:** Display the percentage of male and female employees using a pie chart.
4.  **Experiment 1.4:** Draw a histogram showing salary distribution.
5.  **Experiment 1.5:** Draw a scatter plot showing the relationship between experience and salary.
6.  **Experiment 1.6:** Display the top 10 employees with the highest salary.
7.  **Experiment 1.7:** Find the highest salary in every department.
8.  **Experiment 1.8:** Display employees whose salary is greater than the overall average salary.
9.  **Experiment 1.9:** Calculate the average years of experience in each department.
10. **Experiment 1.10:** Display the distribution of employee ages using a histogram.

## Setup: Import Libraries and Load Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/employee_information_100.csv')

# Display the first few rows of the dataset to verify loading
print("Dataset loaded successfully. Here are the first 5 rows:")
display(df.head())
```

### Experiment 1.1: Average Salary of Each Department

```python
# Calculate the average salary of each department
avg_salary_by_department = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)

print("Average Salary by Department:")
display(avg_salary_by_department)

# Plotting the average salary of each department
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_salary_by_department.index, y=avg_salary_by_department.values, palette='viridis')
plt.title('Average Salary of Each Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.2: Number of Employees in Each Department

```python
# Display the number of employees working in each department
employees_by_department = df['Department'].value_counts().sort_values(ascending=False)

print("Number of Employees in Each Department:")
display(employees_by_department)

# Plotting the number of employees per department
plt.figure(figsize=(10, 6))
sns.barplot(x=employees_by_department.index, y=employees_by_department.values, palette='coolwarm')
plt.title('Number of Employees in Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.3: Percentage of Male and Female Employees

```python
# Calculate the percentage of male and female employees
gender_distribution = df['Gender'].value_counts(normalize=True) * 100

print("Percentage of Male and Female Employees:")
display(gender_distribution)

# Plotting the percentage of male and female employees using a pie chart
plt.figure(figsize=(8, 8))
plt.pie(gender_distribution, labels=gender_distribution.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral'])
plt.title('Percentage of Male and Female Employees')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

### Experiment 1.4: Salary Distribution Histogram

```python
# Draw a histogram showing salary distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['Salary'], bins=10, kde=True, color='purple', edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.5: Relationship Between Experience and Salary

```python
# Draw a scatter plot showing the relationship between experience and salary
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Experience_Years', y='Salary', data=df, hue='Department', size='Salary', sizes=(50, 500), alpha=0.7)
plt.title('Relationship Between Years of Experience and Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.6: Top 10 Employees with Highest Salary

```python
# Display the top 10 employees with the highest salary
top_10_highest_salary = df.nlargest(10, 'Salary')

print("Top 10 Employees with the Highest Salary:")
display(top_10_highest_salary[['Employee_ID', 'Name', 'Department', 'Salary']])
```

### Experiment 1.7: Highest Salary in Every Department

```python
# Find the highest salary in every department
highest_salary_by_department = df.loc[df.groupby('Department')['Salary'].idxmax()]

print("Highest Salary in Every Department:")
display(highest_salary_by_department[['Department', 'Name', 'Salary']].sort_values(by='Department'))
```

### Experiment 1.8: Employees with Salary Greater Than Overall Average

```python
# Calculate the overall average salary
overall_average_salary = df['Salary'].mean()

# Display employees whose salary is greater than the overall average salary
employees_above_average_salary = df[df['Salary'] > overall_average_salary]

print(f"Overall Average Salary: ${overall_average_salary:.2f}")
print("Employees with Salary Greater Than Overall Average:")
display(employees_above_average_salary[['Employee_ID', 'Name', 'Department', 'Salary']].sort_values(by='Salary', ascending=False))
```

### Experiment 1.9: Average Years of Experience in Each Department

```python
# Calculate the average years of experience in each department
avg_experience_by_department = df.groupby('Department')['Experience_Years'].mean().sort_values(ascending=False)

print("Average Years of Experience in Each Department:")
display(avg_experience_by_department)

# Plotting the average years of experience per department
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_experience_by_department.index, y=avg_experience_by_department.values, palette='magma')
plt.title('Average Years of Experience in Each Department')
plt.xlabel('Department')
plt.ylabel('Average Years of Experience')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.10: Distribution of Employee Ages

```python
# Display the distribution of employee ages using a histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=10, kde=True, color='teal', edgecolor='black')
plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
